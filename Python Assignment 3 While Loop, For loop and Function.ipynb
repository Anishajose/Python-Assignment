**Number Guessing Game**

```python
# setup game
import random

# Generate random number between 1 and 10
secret_number = random.randint(1, 10)
# Set a variable attempts to 3,which is the maximum number of guesses allowed.
attempts = 3

# if within the allowed attempts Ask the user to guess the number.
while attempts > 0:
    guess = int(input("Guess the number (between 1 and 10): "))

    # Check if guess is out of range
    if guess < 1 or guess > 10:
        print("Your guess is out of range.Please guess number between 1 and 10.")
        continue

    # Reduce attempt for a valid guess
    attempts -= 1

    # Check the guess
    if guess == secret_number:
      #If the guess is correct, congratulate the user and end the game.
      print("Congratulations! You guessed the correct number!")

      break #break to exit the loop

    elif guess < secret_number:#If the guess is lower than the secret number.
        print("Too low. Try again.")

    else:#If the guess is greater than the secret number.
        print("Too high. Try again.")

else: #user runs out of attempts without guessing the correct number.
    print("Better luck next time!")

    #finding the secret number
    print("The secret number was:", secret_number)
```

**Multiplication Table Generator**

```python
# Take input from the user
number = int(input("Enter the number for which you want the multiplication table: "))

# Generate multiplication table from 1 to 10
for i in range(1, 11): #generates numbers from 1 to 10.
    result = number * i #Calculate the result
    print(number, "x", i, "=", result)# print the tble
```

**BMI Calculator**

```python
# Define the function
def calculate_bmi(weight, height):#passing parameters
    bmi = weight / (height ** 2)#formula for calculation
    return bmi

# Get input from the user
weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

# Calculate BMI using the function
bmi = calculate_bmi(weight, height)

# Display the BMI
print("Your BMI is:", format(bmi, ".2f"))
```
